````
# 2⃣ Get code — self-healing: wipes broken/partial checkouts, then clones or updates
REPO_URL = "https://github.com/hariharanboss/drive-downloader.git"  # @param {type:"string"}
import pathlib, shutil
proj = pathlib.Path("/content/drive-downloader")
healthy = proj.is_dir() and (proj / ".git").exists() and (proj / "app.py").exists()
if proj.exists() and not healthy:
    print(f"🧹 Removing broken checkout at {proj} ...")
    if proj.is_dir():
        shutil.rmtree(proj, ignore_errors=True)
    else:
        proj.unlink(missing_ok=True)
    healthy = False
%cd /content
if not healthy:
    print("📥 Cloning fresh copy ...")
    !git clone $REPO_URL drive-downloader
assert proj.is_dir() and (proj / "app.py").exists(), (
    f"\u274c Clone failed - /content/drive-downloader has no app.py. "
    "Check REPO_URL is correct and the repo is public, then re-run this cell.")
%cd /content/drive-downloader
!git pull --ff-only || echo "\u26a0️ git pull failed (offline?) - continuing with current checkout"
!ls app.py requirements.txt && echo "\u2705 checkout OK"
!pip -q install -r requirements.txt
!pip -q install --upgrade yt-dlp gdown gradio
print("✅ Setup complete - run cell 3.")
````
